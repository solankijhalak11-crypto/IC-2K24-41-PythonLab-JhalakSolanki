# Section D: Problem 3 – K-th Smallest Pairwise Sum

Find the K-th smallest sum made by adding two values from different positions in the list.

```python
import random
import time

# Method 1: Make all pair sums and sort them
def kth_sum_basic(numbers, k):
    pair_sums = []

    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            pair_sums.append(numbers[i] + numbers[j])

    pair_sums.sort()
    return pair_sums[k - 1]


# Count pairs whose sum is at most a given value
def count_small_sums(numbers, limit):
    left = 0
    right = len(numbers) - 1
    count = 0

    while left < right:
        if numbers[left] + numbers[right] <= limit:
            count = count + (right - left)
            left = left + 1
        else:
            right = right - 1

    return count


# Method 2: Use sorting, two pointers and binary search
def kth_sum_fast(numbers, k):
    numbers = sorted(numbers)

    low = numbers[0] + numbers[1]
    high = numbers[-2] + numbers[-1]

    while low < high:
        middle = (low + high) // 2

        if count_small_sums(numbers, middle) >= k:
            high = middle
        else:
            low = middle + 1

    return low
```

```python
random.seed(42)
size = 5000
values = [random.randint(1, 100000) for i in range(size)]
k = 1000000

print("Basic method")
start = time.time()
basic_answer = kth_sum_basic(values, k)
finish = time.time()
print("Answer =", basic_answer)
print("Time:", round(finish - start, 4), "seconds")

print("Faster method")
start = time.time()
fast_answer = kth_sum_fast(values, k)
finish = time.time()
print("Answer =", fast_answer)
print("Time:", round(finish - start, 4), "seconds")

print("Same answer:", basic_answer == fast_answer)
```

## Explanation and Observation

**Step 1 – Basic method**

The basic method creates every possible pair sum, sorts the list, and selects the item at position K. For 5,000 numbers, it creates 12,497,500 sums. This uses a lot of time and memory.

**Step 2 – Binary search idea**

Instead of making every sum, we can ask: how many pair sums are less than or equal to a chosen value? This count helps us narrow down the possible answer.

**Step 3 – Two pointers**

After sorting the numbers, one pointer starts at the left and another at the right. If their sum is within the chosen limit, several pairs can be counted together. Otherwise, the right pointer moves left.

**Step 4 – Search for the answer**

Binary search checks the middle value between the smallest and largest possible pair sums. The count tells us which half to search next.

**Q10. What made the faster method quicker?**

Both ideas helped. It does not create and sort millions of pair sums. Instead, it uses two pointers to count pairs and binary search to find the required value. In the provided run, the basic method took 8.9936 seconds and the faster method took 0.0127 seconds. Times may vary between computers.
