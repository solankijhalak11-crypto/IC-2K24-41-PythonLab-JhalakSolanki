# Section C: Problem 2 – Count Pairs With a Given Sum

Count how many pairs of different positions have values that add up to the target.

```python
import random
import time

# Method 1: Check every pair
def count_pairs_basic(numbers, target):
    count = 0

    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] == target:
                count = count + 1

    return count


# Method 2: Remember values already seen in a dictionary
def count_pairs_fast(numbers, target):
    count = 0
    seen = {}

    for number in numbers:
        needed = target - number

        if needed in seen:
            count = count + seen[needed]

        if number in seen:
            seen[number] = seen[number] + 1
        else:
            seen[number] = 1

    return count
```

```python
print("Basic method")

for size in [5000, 50000]:
    random.seed(42)
    values = [random.randint(1, 20000) for i in range(size)]

    start = time.time()
    answer = count_pairs_basic(values, 20000)
    finish = time.time()

    print("N =", size, "Pairs =", answer)
    print("Time:", round(finish - start, 4), "seconds")


print("Dictionary method")

random.seed(42)
values = [random.randint(1, 20000) for i in range(50000)]

start = time.time()
answer = count_pairs_fast(values, 20000)
finish = time.time()

print("N = 50000, Pairs =", answer)
print("Time:", round(finish - start, 4), "seconds")
```

## Observations

**Q7. What happened when N increased by 10 times?**

The basic method became much slower. It checks pairs using two loops, so increasing the list size causes a large increase in comparisons.

**Q8. Compare the two methods for N = 50,000.**

In the provided run, the basic method took 76.2584 seconds and the dictionary method took 0.0213 seconds. The exact time can change on another computer.

**Q9. What made the second method faster?**

The dictionary made the biggest difference. It stores numbers that have already been visited. For each new number, the program checks whether the number needed to reach the target is already stored. This avoids checking every possible pair.
