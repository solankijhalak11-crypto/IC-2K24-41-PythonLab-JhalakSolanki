# Section B: Problem 1 – Sum of Prime Numbers

The task is to find the sum of all prime numbers smaller than a given number. Three methods are shown: checking every divisor, checking up to the square root, and using the Sieve of Eratosthenes.

```python
import time

# Method 1: Check every possible divisor
def prime_sum_basic(limit):
    total = 0

    for number in range(2, limit):
        prime = True

        for divisor in range(2, number):
            if number % divisor == 0:
                prime = False
                break

        if prime:
            total = total + number

    return total


for value, answer in [(10, 17), (30, 129), (100, 1060), (1000, 76127)]:
    result = prime_sum_basic(value)
    print(value, result, "PASS" if result == answer else "FAIL")
```

```python
# Time the basic method with two input sizes
for value in [50000, 100000]:
    start = time.time()
    answer = prime_sum_basic(value)
    finish = time.time()

    print("N =", value, "Sum =", answer)
    print("Time:", round(finish - start, 4), "seconds")
```

## Method 2: Check Divisors Only Up to the Square Root

```python
def prime_sum_sqrt(limit):
    total = 0

    for number in range(2, limit):
        prime = True

        for divisor in range(2, int(number ** 0.5) + 1):
            if number % divisor == 0:
                prime = False
                break

        if prime:
            total = total + number

    return total


for value, answer in [(10, 17), (30, 129), (100, 1060), (1000, 76127)]:
    result = prime_sum_sqrt(value)
    print(value, result, "PASS" if result == answer else "FAIL")

start = time.time()
answer = prime_sum_sqrt(2000000)
finish = time.time()
print("Sum =", answer)
print("Time:", round(finish - start, 4), "seconds")
```

## Method 3: Sieve of Eratosthenes

```python
def prime_sum_sieve(limit):
    if limit <= 2:
        return 0

    prime = [True] * limit
    prime[0] = False
    prime[1] = False

    number = 2
    while number * number < limit:
        if prime[number]:
            for multiple in range(number * number, limit, number):
                prime[multiple] = False
        number = number + 1

    total = 0
    for number in range(2, limit):
        if prime[number]:
            total = total + number

    return total


for value, answer in [(10, 17), (30, 129), (100, 1060), (1000, 76127)]:
    result = prime_sum_sieve(value)
    print(value, result, "PASS" if result == answer else "FAIL")

start = time.time()
answer = prime_sum_sieve(2000000)
finish = time.time()
print("Sum =", answer)
print("Time:", round(finish - start, 4), "seconds")
```

It was faster because it only checks divisors up to the square root of each number. It still checks numbers one by one, so large inputs can take time.
